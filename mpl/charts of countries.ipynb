```python
import matplotlib.pyplot as plt

labels = ["Service", "Industry", "Agriculture", "Tourist"]
values = [854.8, 408.0, 83.2, 65.2]
colors = ["#08DCDF", "#4BB13F", "#9016A3", "#CAE11D"]
explode = (0.08, 0.02, 0.02, 0.02)

fig, ax = plt.subplots(figsize=(8, 6.5))

wedges, texts, autotexts = ax.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=140,
    colors=colors,
    explode=explode,
    pctdistance=0.75,
    wedgeprops={"edgecolor": "white", "linewidth": 1.5}
)

for autotext in autotexts:
    autotext.set_color("white")
    autotext.set_fontweight("bold")
    autotext.set_fontsize(10)

ax.set_title("Turkiye Income Breakdown (2025 - 4 Categories)", fontsize=13, fontweight="bold", pad=16)

#  (TÜİK 2025 GSYH)
fig.text(
    0.5, 0.03,
    "Total GDP  $1,411.2B",
    ha="center", va="center", fontsize=10, fontweight="bold", color="#1F2937"
)

plt.subplots_adjust(bottom=0.12)
plt.tight_layout(rect=[0, 0.06, 1, 1])
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

countries = ["USA" , "Souviet Union","Great Britain", "Poland","Turkiye"]
country_colors = ["#1D4ED8", "#EE4343", "#23A509", "#D94C06", "#C80F0F"]
gold = [1232, 473, 330, 80, 41]
silver = [1013, 376, 341, 103, 29]
bronze = [892, 361, 350, 152, 41]

x = np.arange(len(countries))
width = 0.25
fig, ax = plt.subplots(figsize=(11, 6))

rects1 = ax.bar(x - width, gold, width, label="Gold", color="#EAB308")
rects2 = ax.bar(x, silver, width, label="Silver", color="#94A3B8")
rects3 = ax.bar(x + width, bronze, width, label="Bronze", color="#B45309")

ax.set_title("All-Time Olympics Medal Count: Top 3 vs Poland & Turkiye", fontsize=13, fontweight="bold", pad=15)
ax.set_ylabel("Number of Medals", fontsize=10)
ax.set_xticks(x)
ax.set_xticklabels(countries, fontsize=10, fontweight="bold")
for tick, col in zip(ax.get_xticklabels(), country_colors):
    tick.set_color(col)
    
ax.set_ylim(0, max(gold) * 1.15)
ax.legend(frameon=False, fontsize=10)

ax.bar_label(rects1, padding=3, fontsize=8.5, fontweight="bold")
ax.bar_label(rects2, padding=3, fontsize=8.5, fontweight="bold")
ax.bar_label(rects3, padding=3, fontsize=8.5, fontweight="bold")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(axis="y", linestyle="--", alpha=0.35)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
